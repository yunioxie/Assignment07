## 09

답 : p는 빨간색이고 삽입 노드의 부모 p^2가 빨간색이면 수선을 해야 한다. p에서 p^6까지 수선이 계속 되려면 모든 노드에서 부모의 형제 노드가 레드여야 한다. p,p^2,p^2의 형제가 레드이면 p^2가 블랙이 되고 p^3은 레드다. p^3,p^4,p^4의 형제가 레드면 p^4가 블랙, p^5는 레드다. p^5,p^6,p^6의 형제가 레드면 p^6이 블랙이 되고 p^5는 레드다. 이런 방식으로 루트까지 올라가서 루트가 레드면 RB트리의 성질에 따라 루트는 블랙이 된다.
